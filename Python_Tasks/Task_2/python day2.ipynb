## Q1. Find the Shortest Word in a Sentence

```python
sentence = "Python makes coding easy"
words = sentence.split(" ")
smallest_word = words[0]
for word in words:
    if len(word)<len(smallest_word):
        smallest_word = word
print(smallest_word)
```

```python
sentence_1 = "Python makes coding useful"
words = sentence_1.split(" ")
large_word = words[0]
final = [large_word]
for word in words[1:]:
    if len(word) > len(large_word):
        large_word = word
        final = [word]  
    elif len(word) == len(large_word):
        final.append(word) 
print(final[0])  
```

## Q6. Count Vowels in a String

```python
word = "Education"
vowels = "aeiouAEIOU"
count = 0
for i in word:
    if i in vowels:
        count += 1
print(count)
```

## Q7. Count Consonants in a String

```python
word = "Python 123!"
vowels = "aeiouAEIOU"
count = 0
for i in word:
    if i.isalpha() and i not in vowels:
        count += 1
print(count)
```

## Q8. Find the Most Frequent Character

```python
string = "banana"
char_counts = {}

for char in string:
    if char in char_counts:
        char_counts[char] += 1
    else:
        char_counts[char] = 1

max_char = ""
max_count = 0
for char,count in char_counts.items():
    if count > max_count:
        max_count = count
        max_char = char
print(max_char)
```

## Q9. Find the Least Frequent Character

```python
string = "success"
char_counts = {}
for char in string:
    if char in char_counts:
        char_counts[char] += 1
    else:
        char_counts[char] = 1
least_frequent_char = min(char_counts, key=char_counts.get)
print(least_frequent_char)
```

## Q10. Find All Characters Appearing More Than Once

```python
string = "programming"
char_counts = {}
res = []
for char in string:
    if char in char_counts:
        char_counts[char] += 1
    else:
        char_counts[char] = 1
max_count = 0
for char,count in char_counts.items():
    if count > max_count:
        max_count = count
        res = [char]  
    elif count == max_count:
        res.append(char)
print(res)
```

## Q15. Create a Dictionary of Numbers and Their Even/Odd Status

```python
numbers = [1, 2, 3, 4, 5]
result = {}
for num in numbers:
    if num % 2 == 0:
        result[num] = "Even"
    else:
        result[num] = "Odd"
print(result)
```

## Q16. Find the Most Frequent Word in a Sentence

```python
sentence = "Python is easy and Python is powerful"
words = sentence.split()
freq = {}
for word in words:
    if word in freq:
        freq[word] += 1
    else:
        freq[word] = 1
max_count = 0
result = ""
for word, count in freq.items():
    if count > max_count:
        max_count = count
        result = word
print(result)
```

## Q17. Find All Words Appearing More Than Once
